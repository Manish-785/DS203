```python
import librosa
import matplotlib.pyplot as plt

# Convert the MFCC DataFrame to a NumPy array
mfccs = df.to_numpy()

# Calculate the delta (first derivative) of the MFCCs
delta_mfcc = librosa.feature.delta(mfccs,order=3)

# Plot the delta MFCCs
plt.figure(figsize=(10, 6))
librosa.display.specshow(delta_mfcc, cmap='coolwarm', x_axis='time', sr=44100, hop_length=512)
plt.colorbar(label='Delta MFCC Coefficients')
plt.title('Delta MFCCs (First Derivative)')
plt.ylabel('MFCC Coefficients')
plt.xlabel('Time (s)')
plt.show()
```

```python
from sklearn.decomposition import PCA
import seaborn as sns
from sklearn.preprocessing import StandardScaler

# Standardize the data
scaler = StandardScaler()
mfcc_data_standardized = scaler.fit_transform(df.T)
# Apply PCA to reduce the MFCCs to 2 dimensions
pca = PCA(n_components=15)
mfcc_pca = pca.fit_transform(mfcc_data_standardized)  # Transpose to have frames as rows

# Create a 2D scatter plot
plt.figure(figsize=(8, 6))
sns.scatterplot(x=mfcc_pca[:, 0], y=mfcc_pca[:, 1], palette='viridis')

plt.title('PCA of MFCCs (2D)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
from sklearn.manifold import TSNE

# Apply t-SNE for visualization
tsne = TSNE(n_components=2, random_state=42)
mfcc_data_tsne = tsne.fit_transform(mfcc_data_standardized)

# Plot the clusters
plt.figure(figsize=(10, 6))
sns.scatterplot(x=mfcc_data_tsne[:, 0], y=mfcc_data_tsne[:, 1], palette='viridis')
```

```python

pca = PCA()
pca.fit(df.T)

# Calculate explained variance
explained_variance = pca.explained_variance_ratio_
cumulative_explained_variance = np.cumsum(explained_variance)

# Plot elbow diagram
plt.figure(figsize=(10, 6))
plt.plot(range(1, len(cumulative_explained_variance) + 1), cumulative_explained_variance, marker='o', linestyle='--')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Elbow Diagram for PCA')
plt.axhline(y=0.95, color='r', linestyle='--')  # Optional: Add a horizontal line at 90% explained variance
plt.grid(True)
plt.show()
```
