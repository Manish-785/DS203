# Load and Prepare Data
Load the data from 'advanced_mfcc_features.csv' and prepare it by reshaping each song's features into a single vector.

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data, skipping the first row
data = pd.read_csv('advanced_mfcc_features.csv', header=0)

# Prepare the data by reshaping each song's features into a single vector
n_songs = 116         # Total number of songs
features_per_song = 36 # Number of features (columns) per song
time_frames = 117     # Number of time frames (rows) per song

# Initialize an empty list to store each song's 4212-dimensional feature vector
song_features = []

for i in range(n_songs):
    # Extract the 36 columns for the current song
    song_data = data.iloc[1:, i * features_per_song:(i + 1) * features_per_song]
    
    # Reshape the song data (117 rows x 36 columns) into a single vector
    song_vector = song_data.values.flatten()
    song_features.append(song_vector)

# Convert to a DataFrame for clustering
songs_df = pd.DataFrame(song_features)

# Normalize the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(songs_df)
songs_df.dropna(inplace=True)
songs_df
# # Apply KMeans Clustering
# n_clusters = 5  # set the desired number of clusters
# kmeans = KMeans(n_clusters=n_clusters, random_state=42)
# labels = kmeans.fit_predict(scaled_data)

# # (Optional) Visualize the Clusters using PCA
# from sklearn.decomposition import PCA
# pca = PCA(n_components=2)
# reduced_data = pca.fit_transform(scaled_data)

# plt.figure(figsize=(10, 7))
# sns.scatterplot(x=reduced_data[:, 0], y=reduced_data[:, 1], hue=labels, palette='viridis')
# plt.title("K-means Clustering of Songs")
# plt.xlabel("PCA Component 1")
# plt.ylabel("PCA Component 2")
# plt.legend(title="Cluster")
# plt.show()
```

# Reshape Song Features
Reshape the song data (116 rows x 36 columns) into a single vector for each song and store these vectors in a DataFrame.

```python
# Reshape the song data (116 rows x 36 columns) into a single vector for each song and store these vectors in a DataFrame

# Initialize an empty list to store each song's 4176-dimensional feature vector
song_features = []

for i in range(n_songs):
    # Extract the 36 columns for the current song
    song_data = data.iloc[:, i * features_per_song:(i + 1) * features_per_song]
    
    # Reshape the song data (116 rows x 36 columns) into a single vector
    song_vector = song_data.values.flatten()
    song_features.append(song_vector)

# Convert to a DataFrame for clustering
songs_df = pd.DataFrame(song_features)
```

# Apply KMeans Clustering
Apply KMeans clustering to the normalized data and set the desired number of clusters.

```python
# Apply KMeans Clustering

# Set the desired number of clusters
n_clusters = 5

# Apply KMeans clustering to the normalized data
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
labels = kmeans.fit_predict(scaled_data)

# (Optional) Visualize the Clusters using PCA
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
reduced_data = pca.fit_transform(scaled_data)

plt.figure(figsize=(10, 7))
sns.scatterplot(x=reduced_data[:, 0], y=reduced_data[:, 1], hue=labels, palette='viridis')
plt.title("K-means Clustering of Songs")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend(title="Cluster")
plt.show()
```

# Visualize Clusters using PCA
Use PCA to reduce the dimensionality of the data and visualize the clusters using a scatter plot.

```python
# Visualize Clusters using PCA
from sklearn.decomposition import PCA

# Reduce the dimensionality of the data to 2 components using PCA
pca = PCA(n_components=2)
reduced_data = pca.fit_transform(scaled_data)

# Create a scatter plot to visualize the clusters
plt.figure(figsize=(10, 7))
sns.scatterplot(x=reduced_data[:, 0], y=reduced_data[:, 1], hue=labels, palette='viridis')
plt.title("K-means Clustering of Songs")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend(title="Cluster")
plt.show()
```
