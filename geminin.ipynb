```python
def generate_questionnaire(row):
    condition_name = row['Name']
    symptoms = row['Symptoms']
    etiology = row['Etiology']

    prompt = f"""
    Create a detailed questionnaire for diagnosing {condition_name}. The questionnaire should be in a multiple-choice format and cover the following aspects:

    1. **Symptoms**: Based on the symptoms provided, create questions that help identify the presence and severity of each symptom.
       - Symptoms: {symptoms}

    2. **Etiology**: Include questions that explore the possible causes and risk factors associated with the condition.
       - Etiology: {etiology}

    Each question should have multiple-choice answers with options.

    Ensure the questions are clear, concise, and cover all relevant aspects to accurately diagnose {condition_name}.
    """

    data = {
        "contents": [
            {
                "parts": [
                    {
                        "text": prompt
                    }
                ]
            }
        ]
    }

    # Make the POST request
    response = requests.post(url, headers=headers, json=data)
    # print(response.json())
    try:
        text = response.json()['candidates'][0]['content']['parts'][0]['text']
    except KeyError:
        text = "Error: Missing expected keys in the response."
    # text = response.json()['candidates'][0]['content']['parts'][0]['text']
    # Assuming 'text' is a string variable containing the text with Markdown syntax
    text_with_new_lines = text.replace('\\n', '\n')

    # Remove common Markdown syntax
    cleaned_text = re.sub(r'(\*\*|##|#|_|[*])', '', text_with_new_lines)

    return cleaned_text
    # return response.json()
```

```python

df['Questionnaire'] = None

for index, row in df.iterrows():
    if index % 10 == 0:
        print(f"Processing row {index}...")
    if df['Questionnaire'][index] is not None:
        continue
    questionnaire = generate_questionnaire(row)
    df['Questionnaire'][index] = questionnaire
    
df.to_csv('questionnaires.csv', index=False)
```
